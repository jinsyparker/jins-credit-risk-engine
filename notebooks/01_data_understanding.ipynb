# 01 Data Understanding

This notebook performs an initial review of the LendingClub loan dataset.

## 1. Import Libraries

```python
# Import libraries for data loading and analysis
import pandas as pd
import numpy as np
```

## 2. Load the LendingClub Data

```python
# Set the path to the raw LendingClub CSV file
data_path = "../data/raw/raw_data.csv"

# Load the dataset into a pandas DataFrame
df = pd.read_csv(data_path, low_memory=False)

# Confirm that the data loaded successfully
print("Dataset loaded successfully.")
```

## 3. Dataset Shape

```python
# Display the number of rows and columns
print("Dataset shape:")
print(df.shape)

print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]:,}")
```

### Milestone 1 Analysis: Dataset Size

**Observation**  
The raw LendingClub dataset contains many rows and a wide set of borrower, loan, payment, hardship, settlement, and credit-history variables.

**Interpretation**  
A wide dataset is useful for exploration, but not every column is appropriate for a Probability of Default model. Some fields may be unavailable at loan origination or may describe events after repayment performance is known.

**Modeling Implication**  
The first PD model should use only variables that would be known before or at the time of the lending decision.

**Decision**  
Use this notebook for understanding the raw data only. Feature eligibility and leakage removal will be handled in the cleaning notebook.

## 6. Data Types

```python
# Show the data type of each column
df.dtypes
```

### Milestone 1 Analysis: Columns and Data Types

**Observation**  
The dataset includes a mix of numeric, categorical, date-like, and percentage-formatted fields.

**Interpretation**  
Columns stored as text may still contain useful numeric information. Examples include interest rate, revolving utilization, term, and employment length.

**Modeling Implication**  
Before modeling, important text-formatted numeric fields should be converted into true numeric variables so that machine learning algorithms can use them correctly.

**Decision**  
Convert key fields in the cleaning notebook and avoid making modeling decisions directly from raw data types.

## 7. Missing Value Summary

```python
# Count missing values in each column
missing_count = df.isna().sum()

# Calculate the percentage of missing values in each column
missing_percentage = (missing_count / len(df)) * 100

# Create a summary table for missing values
missing_values = pd.DataFrame({
    "missing_count": missing_count,
    "missing_percentage": missing_percentage
})

# Sort columns by missing percentage from highest to lowest
missing_values = missing_values.sort_values(by="missing_percentage", ascending=False)

# Display the missing value summary table
missing_values
```

### Milestone 1 Analysis: Missing Values

**Observation**  
The missing value summary identifies which variables have incomplete data and how severe the missingness is.

**Interpretation**  
Missing values are common in credit datasets because some fields only apply to certain borrowers, loan types, hardship cases, joint applications, or later loan events.

**Modeling Implication**  
High-missingness variables may need to be excluded, imputed, or converted into missingness indicators. Missingness should also be reviewed for leakage because some fields may be missing or populated based on future loan behavior.

**Decision**  
Save the missing value table as a Milestone 1 artifact and revisit it during feature selection.

## 8. Loan Status Counts

```python
# Show the count of each loan status category
df["loan_status"].value_counts(dropna=False)
```

### Milestone 1 Analysis: Loan Status and Target Definition

**Observation**  
`loan_status` contains multiple loan outcome categories, not all of which are suitable for a first binary default model.

**Interpretation**  
For a clean Probability of Default target, the first version should focus on completed or clearly defaulted outcomes. Ambiguous or still-active statuses can introduce noise into the target.

**Modeling Implication**  
The target variable should be defined carefully before any modeling. A poor target definition can make model performance misleading.

**Decision**  
In the cleaning notebook, define `default_flag` using `Fully Paid = 0`, `Charged Off = 1`, and `Default = 1`, and exclude other statuses for version 1.

## 9. Save Missing Value Summary

```python
# Set the output path for the missing value summary
output_path = "../reports/tables/missing_values_v1.csv"

# Save the missing value summary table as a CSV file
missing_values.to_csv(output_path, index=True)

print(f"Missing value summary saved to: {output_path}")
```
